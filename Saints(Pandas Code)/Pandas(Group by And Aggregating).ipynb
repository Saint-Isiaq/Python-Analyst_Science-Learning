### Groupby fn()
             multiple ways to run the group by aggregates

```python
group_by_frame = df.groupby('Base Flavor')
```

### Fns() to use with the Group_by_frame
      different aggregates fn()
## Using sum()

```python
group_by_frame.sum()
```

### Using 
    count , min , max and sum FN() 
### avec Group_by

```python
df.groupby('Base Flavor').count()
```

```python
df.groupby('Base Flavor').min()
```

```python
df.groupby('Base Flavor').max()
```

```python
df.groupby('Base Flavor').sum()
```

### Using 
       Aggregate Fn() Avec min, max ,mean, count, sum fn()
## Avec Groupby fn()

```python
df.groupby('Base Flavor').agg({'Flavor Rating':['mean','sum','max','count'], 'Texture Rating':['sum','min','mean','max']})
```

### Grouping Multiple Columns

```python
df.groupby(['Base Flavor','Liked']).sum()

#OR
```

```python
df.groupby(['Base Flavor','Liked']).agg({'Flavor Rating':['mean','sum','max','count'], 'Texture Rating':['sum','min','mean','max']})
#or
```
