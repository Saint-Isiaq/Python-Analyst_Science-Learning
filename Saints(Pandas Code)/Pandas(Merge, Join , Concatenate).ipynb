### Using
  .merge(df, how = 'outer')
### Defaults : full outer join; Data that both DF have in them
### Note: 
   NaN means: Not a nUMBER; when a common match can't be found in both df's data

```python
df1.merge(df2, how = 'outer')
```

### how = 'Left'  (left outer join)
    all data from the left df , then brings in matching data from the other df

```python
df1.merge(df2, how = 'left')
```

### how = 'right'
    exact opposite of the how = 'left' ,
    all data from the right table then matching from the left

```python
df1.merge(df2, how = 'right')
```

### how = 'cross'  : Cross Join
    takes each value in the left df and compares with all the value on the right df
    and repeats it for the next value 

```python
df1.merge(df2, how = 'cross')
```

### Using Joins

```python
df1.join(df2, on = 'FellowshipID', how = 'outer', lsuffix = '_Left', rsuffix = '_Right')

#OR
```

### To make join work as merge
  Calling it in another df

```python
df5 = df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'), 
                                         lsuffix = '_Left', rsuffix = '_Right',how = 'outer')
df5
```

### Concatenate
    this puts a df on top of each other, 
         instead of beside one another like the merge and join

```python
#pd.concat([df1,df2])
#OR
pd.concat([df1,df2],join = 'inner')  #OR join = 'outer'
```

### Concat by Axis
   """Axis = 1 ; column based concat
      Axis = 0 ; index/row based concat
   """

```python
pd.concat([df1,df2],join = 'outer',axis = 1)
```
