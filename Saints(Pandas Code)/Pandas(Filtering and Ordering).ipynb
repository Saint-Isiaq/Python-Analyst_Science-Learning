### Filtering by coumns

### Filtering by Specifying Columns

```python
specific_countries = ['Brazil','Indonesia','Nigeria','Russia']

df[df['Country'].isin(specific_countries)]
```

### Filtering By String Value

```python
df[df['Country'].str.contains('United')]
```

```python
#Or 
df[df['Continent'].str.contains('Asia')]  #brings_All_countries in asia contained in the file
```
