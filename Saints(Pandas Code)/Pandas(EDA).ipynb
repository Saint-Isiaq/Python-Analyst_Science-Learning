### Visualizing correlation Using a heatmap
    and size re-correction

```python
# sns.heatmap(df.corr(numeric_only = True),annot = True)
# OR
sns.heatmap(df.corr(numeric_only = True),annot = True)

plt.rcParams['figure.figsize'] = (20,7)

plt.show()
```

### Grouping columns together
     to look at the data a lil closer
using .groupby Fn()

```python
df.groupby('Continent').max().sort_values(by="2022 Population",ascending = True)
```

```python
df[df['Continent'].str.contains('Oceania')]
```

### Showing graphs for the groupby

```python
df2 = df.groupby('Continent').max().sort_values(by="2022 Population",ascending = True)
```

```python
df2.plot()
```
